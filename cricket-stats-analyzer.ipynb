```python
import pandas as pd
import numpy as np

df=pd.read_csv("cricket_stats.csv")
df
```

```python
df['average']=df['average'].fillna(df['average'].mean())
df['matches']=df['matches'].fillna(df['matches'].mean())
df['runs']=df['runs'].fillna(df['runs'].mean())
df['centuries']=df['centuries'].fillna(0)
```

# Q1 Who has the highest average?

```python

top_avg_player=df.loc[df['average'].idxmax()]['player']
print(f"Best average: {top_avg_player}")
```

# Q2 Which team has the most total runs combined?

```python
most_total_runs=df.groupby('team')['runs'].sum()
team_with_most_runs=most_total_runs.idxmax()
print(f"Team with most runs: {team_with_most_runs}")
```

# Q3 Who has the best strike rate among players with more than 40 matches?

```python
filteration=df[df['matches']>40]
best_strike_rate=filteration.loc[filteration['strike_rate'].idxmax(),'player']
print(f"Best strike rate (40+ matches): {best_strike_rate}")
```

#Q4 What's the average number of centuries across all players?

```python
avg_num_of_centuries=df['centuries'].mean()
print(f"Average number of centuries: {avg_num_of_centuries:.2f}")
```

#Q5 Rank all players by runs, highest to lowest

```python
ranking=df.sort_values(by='runs',ascending=False)
print(ranking[['player','runs']])
```

```python
print("=== Cricket Stats Report ===")
print(f"Best average: {top_avg_player}") #Q1
print(f"Team with most runs: {team_with_most_runs}") #Q2
print(f"Best strike rate (40+ matches): {best_strike_rate}") #Q3
print(f"Average number of centuries: {avg_num_of_centuries:.2f}") #Q4
print(ranking[['player','runs']]) #Q5
```
